```python
#--------------------------------
# 1.Load the dataset using pandas.
#--------------------------------

import pandas as pd
df = pd.read_csv('Iris.csv')
```

```python
#----------------------------------------------------------------------
# 2.Print the shape, column names, and the first few rows using .head().
#----------------------------------------------------------------------


# shape
print(df.shape)

# coloumns
print(df.columns)

# first 5 row
print(df.head())
```

```python
#------------------------------------------------------
# 3. Use .info() and .describe() for summary statistics.
#------------------------------------------------------

# .info()
print(df.info())

# .describe()
print(df.describe())
```

```python

#------------------------------
# 4.Visualize the dataset:
#------------------------------

import matplotlib.pyplot as plt

# 1. Create a scatter plot to show relationships between features.

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)

# Scatter plots with different colors
plt.scatter(df['SepalLengthCm'], df['SepalWidthCm'], color='blue', label='Sepal')
plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'], color='red', label='Petal')

# Add title and labels
plt.title("Relationship between 4 features")
plt.xlabel("Length")
plt.ylabel("Width")

# Add legend to distinguish colors
plt.legend()
plt.show()
```

```python
# 2.  Use histograms to show value distributions.

plt.figure(figsize=(10, 5))
plt.subplot(1, 4, 1)
plt.hist(df['PetalLengthCm'],bins=10 ,color='red')
plt.title("Petal Length")


plt.subplot(1, 4, 2)
plt.hist(df['PetalWidthCm'],bins=10,color='orange')
plt.title("Petal Width")


plt.subplot(1, 4, 3)
plt.hist(df['SepalLengthCm'],bins=10 ,color='blue')
plt.title("sepal Length")


plt.subplot(1, 4, 4)
plt.hist(df['SepalWidthCm'],bins=10,color='gray')
plt.title("Sepal Width")


# Add spacing between subplots
plt.subplots_adjust(wspace=0.4, hspace=0.4)
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
# 3. Use box plots to identify outliers.

plt.boxplot([df['SepalLengthCm'], df['SepalWidthCm']])
plt.title("Sepal Features Comparison")
plt.xticks([1, 2], ['Sepal Length', 'Sepal Width'])
plt.show()
```

```python
plt.boxplot([df['PetalLengthCm'], df['PetalWidthCm']])
plt.title("Petal Features Comparison")
plt.xticks([1, 2], ['Petal Length', 'Petal Width'])
plt.show()
```
